```python
# specify input and output files here
file_input = '<input file>'
file_output = '<output file>'
```

```python
import pandas as pd

# parse and rename columns
# given format works for day-ahead prices ("Großhandelspreise") downloaded from https://www.smard.de/home/downloadcenter/download-marktdaten
df = pd.read_csv(file_input, sep=';', header=0, names=['time', 'price'], usecols=[0, 2], decimal=',')

# convert time column to proper format
df['time'] = pd.to_datetime(df['time'], format='%d.%m.%Y %H:%M')
```

```python
from datetime import timezone, timedelta

# in the time series from smard (see above), date/time is affected by
# daylight saving time so that from the last Sunday in March to the
# last Sunday in October, date/time is shifted by one hour. This might
# differ/not be the case for different data sources.

# CET in python uses daylight saving time, so we define our own CET
# as UTC+1 here
cet_fixed = timezone(timedelta(hours=1), name="CET")

# setting correct time zone for input data
df['time'] = df['time'].dt.tz_localize("Europe/Berlin", ambiguous="infer")

# converting to CET (UTC+1)
df['time'] = df['time'].dt.tz_convert(cet_fixed)
```

```python
# convert time to iso format
df['time'] = pd.Series(df['time'].dt.to_pydatetime()).map(lambda datetime: datetime.isoformat())
```

```python
# save to output file
df.to_csv(file_output, sep=',', index=False)
```
